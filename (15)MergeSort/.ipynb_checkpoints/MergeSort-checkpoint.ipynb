```python
# Merge Sort : We have an unsorted List. we break that list in half until the only thing
# that we have left are lists that only have 1 item in them.Which mean by defination all the list are sorted.
# Now we will take two of the items and create a new list that's sorted, we will do same with next two list, then next two, 
# until we have 4 list instead of 8. Then we merge and sort left 2 lists, then right 2 lists which will give us 2 lists of 4
# items each. Now we will take these 2 lists and merge and sort.

# Merge function which is called as helper function does'nt do all the things which are required for merge sort.
# It takes two sorted list and it takes the values from these and puts them into combined list.

# For the 2 lists, we create for loop for each of the List. We will compare i and j, whichever one is lowest we will add to
# new combined list.
   0   1   2   3      0   1   2   3
# |1 | 3 | 7 | 8 |   |2 | 4 | 5 | 6 |
#  i                  j

# Now 
# 1. We break list in half - using recursion
# 2. Base Case: when len(the_list) is 1
# 3. Use merge() to put lists together.

# Big 0
# So we break the list into number of items present. We have 8 new list with each having length of 1.
# So number of items stored in memory have been doubled. So that means this is O(n) for space complexity.
# Now for list of 8 items, we break it in half 3 times to get single item list, which is Log2 8 = 3
# So breaking list apart has O(log n) and putting it back together because of havinh a loop through each item with
# merge function is O(n). So putting it together becomes O(n log n). This is the most efficient that you can make sorting 
# algorithm.
```

```python
def merge(list1, list2):
    combined = [] #Empty list
    i = 0
    j = 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            combined.append(list1[i])
            i+=1
        else:
            combined.append(list2[j])
            j+=1
    while i < len(list1):
        combined.append(list1[i])
        i+=1
    while j < len(list2):
        combined.append(list2[j])
        j+=1
    return combined

# print(merge([1,2,7,8],[3,4,5,6]))

def merge_sort(my_list):
    if len(my_list) == 1:
        return my_list
    mid_index = int(len(my_list)/2) #incase we have list with odd numbers which cannot be divided by 2
    left = merge_sort(my_list[:mid_index]) #calling merge sort recursively
    right = merge_sort(my_list[mid_index:])
    
    return merge(left, right)


original_list = [3,1,2,4]
sorted_list = merge_sort(original_list)
print('Original List', original_list)
print('\nSorted list', sorted_list)
```
