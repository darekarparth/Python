```python
# Tree Traversal means, we are going to visit every node in the tree
# Then we are going to take the value and put it in the list and then we will return that list.
# BFS (Breadth First Search)- We start at the top(root) and then we traverse to next row(left child then right child)  and so on.
# DFS (Depth First Search)- We start from bottom left child then come to left parent and again go to right child then go to root then go to bottom left child of 
# right parent then go right parent and then got to bottom right child of right parent.
```

```python
# DFS- Depth first search:: PreOrder
# PreOrder DFS - The order in which we are going to add the numbers to the list will be like : 
# We will start at top and add no. 47, 21 to left, then 18, then come back at 21 and go to right to add 27.
#  Once this is done, we will go to right of 47, 76, then 52, then again 76 and 82.
       #         47
       #       /    \
       #     21      76
       #   /   \    /    \   
       # 18    27  52    82
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
class BinarySearchTree:
    def __init__(self):
        self.root = None #keep none as initially

    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return True
        temp = self.root
        while temp is not None:
            if new_node.value == temp.value:
                return False
            if new_node.value < temp.value:
                if temp.left is None:
                    temp.left = new_node
                    return True
                temp = temp.left
            else:
                if temp.right is None:
                    temp.right = new_node
                    return True
                temp = temp.right

    def contains(self, value):
            if self.root is None:
                return False
            temp = self.root
            while temp is not None:
                if value < temp.value:
                    temp = temp.left
                elif value > temp.value:
                    temp = temp.right
                else:
                    return True
            return False

    def dfs_pre_order(self):
        results = [] #empty list defined
        
        def traverse(current_node):
            results.append(current_node.value)
            if current_node.left is not None:
                traverse(current_node.left)
        
            if current_node.right is not None:
                traverse(current_node.right)
        
        traverse(self.root)
        return results

my_Tree = BinarySearchTree()
my_Tree.insert(47)
my_Tree.insert(21)
my_Tree.insert(76)
my_Tree.insert(18)
my_Tree.insert(27)
my_Tree.insert(52)
my_Tree.insert(82)

print(my_Tree.dfs_pre_order())

```

```python
# DFS- Depth first search: Post-Order
# We are first going to visit 47, then 21 to left, then 18, after that, 18 is going to look right and left
# and there's no node there, and finally it will write it's value to result list. And now we come up to 21 again, it has
# already gone left, so it will go to right to 27. Then 27 is going to look left and write and finally write the value 27 
# to the result list. Now we come back up to 21 again. It has gone left and write, so we will write value 21 to list. 
# And this will bring us back to 47, it has gone left, now it will go right to the 76 node. And here we always staart with going 
# to left first, it will go to 52 node, look lleft and right, and then write value 52 to results list. We come back upto 76, 
# Now it will go right that 82 node will look left and right and then write value 82 to results list. That brings us bact upto 76, 
# As it has gone left and right and can write it's value 76 to results list and that will brings us back to 47 as it has gone
# left and right and now its value can be written to results list.

       #         47
       #       /    \
       #     21      76
       #   /   \    /    \   
       # 18    27  52    82
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
class BinarySearchTree:
    def __init__(self):
        self.root = None #keep none as initially

    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return True
        temp = self.root
        while temp is not None:
            if new_node.value == temp.value:
                return False
            if new_node.value < temp.value:
                if temp.left is None:
                    temp.left = new_node
                    return True
                temp = temp.left
            else:
                if temp.right is None:
                    temp.right = new_node
                    return True
                temp = temp.right

    def contains(self, value):
            if self.root is None:
                return False
            temp = self.root
            while temp is not None:
                if value < temp.value:
                    temp = temp.left
                elif value > temp.value:
                    temp = temp.right
                else:
                    return True
            return False

    def dfs_post_order(self):
        results = []

        def traverse(current_node):
            if current_node.left is not None:
                traverse(current_node.left)
            if current_node.right is not None:
                traverse(current_node.right)
            results.append(current_node.value) 

        traverse(self.root)
        return results
        

my_Tree = BinarySearchTree()
my_Tree.insert(47)
my_Tree.insert(21)
my_Tree.insert(76)
my_Tree.insert(18)
my_Tree.insert(27)
my_Tree.insert(52)
my_Tree.insert(82)

print(my_Tree.dfs_post_order())
```

```python
# DFS- Depth first search: Inorder
# We visit node 47, then we traverse to left 21 then to left 18 again. The 18 will try to go to left 
# but there's nothing there, so we will write 18 to our list, and then it will go to right and there is
# nothing there. Then we come up to 21, write 21 to the list and go to right to 27. 27 will go left, 
# there is nothing, write it's value 27 to list, then go right, and there's nothing. Then we will 
# come back to 21 and then 47. Now we write value 47 to list, then it will go to 76, then it will 
# go to 52, 52 will try to go left, but there's nothing there, then we will write 52 to list then
# 52 will go to right but there is nothing, so we will traverse back to 76. Then we will write 76 to
# list. Then 76 will go to right to 82, 82 will look left, nothing, then write itself to list and go 
# right, nothing. 
       #         47
       #       /    \
       #     21      76
       #   /   \    /    \   
       # 18    27  52    82
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
class BinarySearchTree:
    def __init__(self):
        self.root = None #keep none as initially

    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return True
        temp = self.root
        while temp is not None:
            if new_node.value == temp.value:
                return False
            if new_node.value < temp.value:
                if temp.left is None:
                    temp.left = new_node
                    return True
                temp = temp.left
            else:
                if temp.right is None:
                    temp.right = new_node
                    return True
                temp = temp.right

    def contains(self, value):
            if self.root is None:
                return False
            temp = self.root
            while temp is not None:
                if value < temp.value:
                    temp = temp.left
                elif value > temp.value:
                    temp = temp.right
                else:
                    return True
            return False

    def dfs_inorder(self):
        results = []
        def traverse(current_node):
            if current_node.left is not None:
                traverse(current_node.left)
            results.append(current_node.value)
            if current_node.right is not None:
                traverse(current_node.right)

        traverse(self.root)
        return results
        

my_Tree = BinarySearchTree()
my_Tree.insert(47)
my_Tree.insert(21)
my_Tree.insert(76)
my_Tree.insert(18)
my_Tree.insert(27)
my_Tree.insert(52)
my_Tree.insert(82)

print(my_Tree.dfs_inorder())

```
