```python
# Quick Sort: We start with the pivot point, which is 1st number in the list. Then we will compare each of the number after
# that to pivot item to see it it's greater than or less than.

def swap(my_list, index1, index2):
    temp = my_list[index1]
    my_list[index1] = my_list[index2]
    my_list[index2] = temp

def pivot(my_list, pivot_index, end_index):
    swap_index = pivot_index
    
    for i in range(pivot_index+1, end_index+1):  #for loop is going to go to upto but not including 'end_index+1' value.
        if my_list[i] < my_list[pivot_index]:
            swap_index +=1
            swap(my_list, swap_index, i)
    swap(my_list, pivot_index, swap_index)
    return swap_index

def quick_sort(my_list, left, right):
    if left < right:
        pivot_index = pivot(my_list, left, right) #Storing the index that pivot function returns which is equal to 4
        quick_sort(my_list, left, pivot_index-1) #recursively call quick sort on left. 
        quick_sort(my_list, pivot_index+1, right)
    return my_list        

# my_list = [4,6,1,7,3,2,5]
# print(pivot(my_list, 0, 6))
# print(my_list)   

print(quick_sort([4,6,1,7,3,2,5], 0, 6))

# Big O - O(n log n)
        
```
