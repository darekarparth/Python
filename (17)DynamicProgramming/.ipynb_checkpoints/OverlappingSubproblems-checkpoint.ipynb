```python
# Overlapping Subproblems
# Optimized Substructure
# Fibonacci Sequence 


counter = 0

def fib(n):
    global counter
    counter += 1
    if n == 0 or n == 1:
        return n
    return fib(n-1)+fib(n-2)

n = 7
print(fib(n))
print(counter) #There are 41 function calls that you have to have to find fib(7). 
# This is very inefficient.
```

```python
# Memoization
# Top - Down Approach
memo = [None]*100 #initializing 100 indexes with None
counter = 0
def fib(n):
    global counter
    counter +=1
    if memo[n] is not None:
        return memo[n]
        
    if n == 0 or n == 1:
        return n
        
    memo[n] = fib(n-1)+fib(n-2)
    return memo[n]
n = 35
print(fib(n))
print(counter)
```

```python
# Down - Top approach - More better
counter = 0
def fib(n):
    fib_list = [0, 1]
    global counter
    
    for index in range(2, n+1):
        counter +=1
        next_fib = fib_list[index-1] + fib_list[index-2]
        fib_list.append(next_fib)
    return fib_list[n]

n = 7
print(fib(n))
print(counter)
```
