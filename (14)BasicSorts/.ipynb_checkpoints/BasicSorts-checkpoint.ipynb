```python
# Bubble Sort
def bubble_sort(my_list):
    # The range of loop is length - 1, which is equal to 5. Then we are going to start at 0th position, 
    # and then we are going to decrement it each time by -1.
    for i in range(len(my_list) - 1, 0, -1):
        # The second for loop is going to move forward through list(which is going to run 5 times
        # to do 5 comparisons. And then when we go through this again, i will be 4 and this loop 
        # run 4 times, then 3, then 2)
        for j in range(i):
            if my_list[j] > my_list[j+1]:
                # swapping method
                temp = my_list[j]
                my_list[j] = my_list[j+1]
                my_list[j+1] = temp
    return my_list

my_list = [4,9,5,3,1,8,2,7,6]
    
print(bubble_sort(my_list))
```

```python
# Selection Sort,
# We implement this sort using the index of the list values. We compare the 
# value with the first index with the next minimum value index and swap it.

def selection_sort(my_list):
    for i in range(len(my_list)-1):
        min_index = i
        for j in range(i+1, len(my_list)): #i+1 is start value for our loop, and len(my_list) is end value
            if my_list[j] <  my_list[min_index]:
                min_index = j
        if i != min_index:
            temp = my_list[i]
            my_list[i] = my_list[min_index]
            my_list[min_index] = temp
    return my_list

my_list = [4,9,5,3,1,8,2,7,6]

print(selection_sort(my_list))
            
```

```python
# Insertion Sort.
# We always starts with second item in the list and we compare it to item before it.
# If it's less than the item before we will swap it in correct spot.
# then we go to 3rd item and we compare it to item in 2nd place and so on ...

def insertion_sort(my_list):
    for i in range(1, len(my_list)): #1(index 1) is the start value for our loop, and len(my_list) is end value
        temp = my_list[i]
        j = i - 1
        while temp < my_list[j] and j > -1:
            my_list[j+1] = my_list[j] #moving 9 to place of temp variable 5
            my_list[j] = temp #Moving 5 to position of 9. 
            j -=1 #Once swappong is done. 5 is moved to position j. but 5 is temp variable which is i. so we need to move
            # j to previous position pointing to 4
    return my_list

my_list = [4,9,5,3,1,8,2,7,6]

print(insertion_sort(my_list))

# Insertion Sort Big 0
# O(n^2) for unsorted list
# O(n) for somewhat or mostly  sorted list
        
```
