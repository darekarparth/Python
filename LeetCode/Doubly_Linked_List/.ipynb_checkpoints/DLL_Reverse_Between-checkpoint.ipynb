```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
        self.prev = None

class DoublyLL:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1

    def print_list(self):
        temp = self.head
        # values = []
        while temp is not None:
            # values.append(temp.value)
            print(temp.value)
            temp = temp.next

        # return values

    def append(self, values):
        new_node = Node(values)
        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node
            new_node.next = None
        self.length+=1    

    def pop_last(self): #By LL and DLL 
        # if self.length == 0:
        #         return None
        # temp = self.head #pointers
        # pre = self.head #pointers
        # while temp.next is not None: #Sliding window giving us second last value
        #     pre = temp
        #     temp = temp.next
            
        # self.tail = pre
        # self.tail.next = None
        # self.length -= 1
    
        if self.length == 0:
            return 0
        temp = self.tail
        if self.length == 1 :
            self.head = None
            self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None
            temp.prev = None
        self.length-=1
    
    def prepend(self, value): #By LL and DLL
        # new_node = Node(value)
        # if self.length == 0:
        #     return 0
        # else:
        #     new_node.next = self.head
        #     self.head = new_node
        # self.length+=1

        new_node = Node(value)
        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
        self.length+=1
    
    def pop_first(self): #By LL and DLL
        # if self.length == 0:
        #     return 0
        # else:
        #     temp = self.head
        #     self.head = self.head.next
        #     temp.next = None
        #     temp = None
        # self.length-=1
        if self.length == 0:
             return 0
        pre = self.head #pointer
        if self.length == 1:
            self.head = new_
        else:
            self.head = self.head.next
            self.head.prev = None
            pre.next = None
        self.length-=1    
            

    def get(self, index): #same for Ll and DL
        if index < 0 and index >= self.length:
            return 0
        temp = self.head
        for _ in range(index):
            temp = temp.next
        return temp

    def set_value(self, index, value): #Same for LL and DL
        temp = self.get(index)
        if temp is not None:
           temp.value = value
           return True
        return False


    def insert(self, index, value): #By LL and By DLL
        # if index < 0 and index > self.length:
        #     return False
        # if index == 0:
        #     return self.prepend(value)
        # if index == self.length:
        #     return self.append(value)
        # new_node = Node(value)
        # temp = self.get(index-1)
        # new_node.next = temp.next
        # temp.next = new_node
        if index < 0 and index > self.length:
            return False
        if index == 0:
            return self.prepend(value)
        if index == self.length:
            return self.append(value)
        new_node = Node(value)
        temp = self.get(index-1)
        pre = temp.next
        new_node.next = temp.next
        pre.prev = new_node
        temp.next = new_node
        new_node.prev = temp
        self.length +=1

    def remove(self, index):#By LL and By DLL
        # if index < 0 or index > self.length:
        #     return False
        # if index == 0:
        #     return self.pop_first()
        # if index == self.length - 1:
        #     return self.pop_last()
        # temp = self.get(index-1)
        # pre = temp.next
        # temp.next = pre.next
        # pre.next = None
        # self.length-=1
        if index < 0 or index > self.length:
            return False
        if index == 0:
            return self.pop_first()
        if index == self.length - 1:
            return self.pop_last()
        temp = self.get(index-1)
        pre = temp.next
        temp.next = pre.next
        pre.prev = temp
        self.length-=1

    def reverse(self): #Swapping positions of head and tail first / By Ll
        temp  = self.head
        self.head = self.tail
        self.tail = temp
        after = temp.next #pointer
        before = None #pointer
        for _ in range(self.length):
            after = temp.next #pointer relocation
            temp.next = before #Value Assignement
            before = temp #pointer relocation
            temp = after #pointer relocation
        
    def is_palindrome(self):
        temp = self.head #pointer
        pre = self.tail #pointer
        values_1 = []
        values_2 = []
        while temp and pre is not None:
            values_1.append(temp.value)
            temp = temp.next
            values_2.append(pre.value)
            pre = pre.prev
        if values_1 == values_2:
            return True
        else:
            return False

    # def is_palindrome(self):
    #     if self.length <= 1:
    #         return True
    #     forward_node = self.head
    #     backward_node = self.tail
    #     for i in range(self.length // 2):
    #         if forward_node.value != backward_node.value:
    #             return False
    #         forward_node = forward_node.next
    #         backward_node = backward_node.prev
    #     return True
    
    def reverse_dll(self):
        current = self.head
        while current is not None: #swapping the nodes
            temp = current.prev #pointer
            current.prev = current.next#value assignment
            current.next = temp#pointer moved
            temp = temp.prev
        dummy = self.head
        self.head = self.tail
        self.tail = dummy

    # def reverse_dll(self):
    #     temp = self.head
    #     while temp is not None:
    #         # swap the prev and next pointers of node points to
    #         temp.prev, temp.next = temp.next, temp.prev
            
    #         # move to the next node
    #         temp = temp.prev
            
    #     # swap the head and tail pointers
    #     self.head, self.tail = self.tail, self.head



    def partition_List(self, x):
        if not self.head:
            return None
        dummy_1 = Node(0)
        dummy_2 = Node(0)
        prev1 = dummy_1
        prev2 = dummy_2
        current = self.head
        while current is not None:
            if current.value < x:
                prev1.next = current
                current.prev = prev1
                prev1 = current
            else:
                prev2.next = current
                current.prev = prev2
                prev2 = current
            current = current.next

            #joining both nodes
        prev1.next = dummy_2.next
        if dummy_2.next is not None:
            forward = dummy_2.next
            forward.prev = prev1
        prev2.next = None
        self.head = dummy_1.next 
        self.head.prev = None


    def reverse_between(self, start_index, end_index):
        if self.length <= 1 or start_index == end_index:
            return
        dummy_node = Node(0)
        dummy_node.next = self.head
        self.head.prev = dummy_node
        # Traverse to the node just before the start_index
        prev1 = dummy_node #pointer
        for _ in range(start_index):
            prev1 = prev1.next
        current = prev1.next  # current points to the first node in the segment to reverse
        for _ in range(end_index - start_index):
            node_to_move = current.next
            current.next = node_to_move.next
            if node_to_move.next:
                node_to_move.next.prev = current
            node_to_move.next = prev1.next
            prev1.next.prev = node_to_move
            prev1.next = node_to_move
            node_to_move.prev = prev1

        self.head = dummy_node.next
        self.head.prev = None
            
            
                
                
           
d_ll = DoublyLL(1)
d_ll.append(2)
d_ll.append(3)
d_ll.append(4)
d_ll.append(5)
d_ll.print_list()
print()
# d_ll.pop_last()
# d_ll.print_list()
# print()
# d_ll.prepend(10)
# d_ll.print_list()
# print()
# d_ll.pop_first()
# d_ll.print_list()
# print()
# print(d_ll.get(2).value)
# print()
# d_ll.set_value(0,5)
# d_ll.print_list()
# print()
# d_ll.insert(3,10)
# d_ll.print_list()
# print()
# d_ll.remove(3)
# d_ll.print_list()
# print()
# d_ll.reverse()
# d_ll.print_list()
# print(d_ll.is_palindrome())
# d_ll.reverse_dll()
# d_ll.print_list()
# d_ll.partition_List(3)
# d_ll.print_list()
d_ll.reverse_between(1,3)
d_ll.print_list()
```
