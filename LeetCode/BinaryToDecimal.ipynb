```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1


    def append(self, value):
        new_node = Node(value)
        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

    def print_list(self):
        if self.head is None:
            print("List is empty")
        else:
            temp = self.head
            values = [] #empty Array
            while temp is not None:
                values.append(str(temp.value))
                temp = temp.next
            print("->".join(values))

    def binary_to_decimal(self):
        num = 0 #new variable num
        current = self.head #pointer
        while current is not None:
            num = num * 2 + current.value  
            #0*2+1=1
            #1*2+0=2
            #2*2+1=5
            #5*2+1=11
            #Number = 11
            current = current.next
        return num
            
        


my_LL = LinkedList(1)
my_LL.append(0)
my_LL.append(1)
my_LL.append(1)

print("Test case 1 linked list:")
my_LL.print_list()
result = my_LL.binary_to_decimal()
try:
    assert result == 11
    print("Test case 1 passed, returned:", result)
except AssertionError:
    print("Test case 1 failed, returned:", result)
print("-" * 40)
```
