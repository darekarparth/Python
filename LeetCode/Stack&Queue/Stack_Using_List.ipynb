```python
class Stack:
    def __init__(self):
        self.stack_list = []

    def print_list(self):
        for i in range(len(self.stack_list)-1, -1, -1):
            print(self.stack_list[i])

    def is_empty(self):
        return len(self.stack_list) == 0

    def peek(self):
        if self.is_empty():
            return None
        else:
            return self.stack_list[-1]

    def size(self):
        return len(self.stack_list)
    
    def push(self, value):
        self.stack_list.append(value)

    def pop(self):
        if self.is_empty():
            return None
        else:
            return self.stack_list.pop()


def is_balanced_parentheses(parentheses):
    stack = Stack()
    for p in parentheses:
        if p == '(':
            stack.push(p)
        elif p == ')':
            if stack.is_empty() or stack.pop() != '(':
                return False
    return stack.is_empty()



def reverse_string(my_string):
    stack = Stack()
    reversed_string = ""
    for char in my_string:
        stack.push(char)
    stack.print_list()
    while not stack.is_empty():
        reversed_string += stack.pop()

    return reversed_string
    

def sort(my_stacked):
    addstack = Stack()
    while not my_stacked.is_empty():
        temp = my_stacked.pop()
        while not addstack.is_empty() and addstack.peek() > temp:
            my_stacked.push(addstack.pop())
        addstack.push(temp)

    while not addstack.is_empty():
        my_stacked.push(addstack.pop())
        

my_stack = Stack()
my_stack.push(4)
my_stack.push(2)
my_stack.push(3)
my_stack.push(5)
my_stack.push(1)
my_stack.print_list()
print()

# my_stack.pop()
# my_stack.print_list()

def test_is_balanced_parentheses():
    try:
        assert is_balanced_parentheses('((()))') == True
        print('Test case 1 passed')
    except AssertionError:
        print('Test case 1 failed')


my_string = "Lamborghini"
print(reverse_string(my_string))


sort(my_stack)
my_stack.print_list()
```
